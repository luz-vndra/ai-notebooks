# Tokenization

- install `tiktoken` with `uv add tiktoken` (`tiktoken` works only for OpenAI model names)

```python
import tiktoken

encoding = tiktoken.encoding_for_model("gpt-4")

tokens = encoding.encode("Vegeta is the prince of saiyans.")

print(tokens)
```

```python
for token in tokens:
    print(token, encoding.decode([token]))
```

```python
encoding.decode([1, 42826, 500, 1])
```

# (Illusion of) Memory

```python
# setup llama3.2 

from openai import OpenAI

llama = OpenAI(base_url='http://localhost:11434/v1', api_key='ollama')
```

### A message to OpenAI client is a list of dicts

```python
messages = [
    {"role": "system", "content": "You are a helpful assistant"},
    {"role": "user", "content": "Hi! I'm Vulz!"}
]

response = llama.chat.completions.create(model="llama3.2", messages=messages)
response.choices[0].message.content
```

### A follow-up question

```python
messages = [
    {"role": "system", "content": "You are a helpful assistant"},
    {"role": "user", "content": "What's my name?"}
]

response = llama.chat.completions.create(model="llama3.2", messages=messages)
response.choices[0].message.content
```

**This call to the LLM is stateless -- so the previous answer is not recorded. It is up to the developer to preserve the response and sequence it each time in the chat**

### The sequencing

```python
messages = [
    {"role": "system", "content": "You are a helpful assistant"},
    {"role": "user", "content": "Hi! I'm Vulz!"}
]

response = llama.chat.completions.create(model="llama3.2", messages=messages)
chat_response = response.choices[0].message.content

print(chat_response)

messages = [
    {"role": "system", "content": "You are a helpful assistant"},
    {"role": "user", "content": "Hi! I'm Vulz!"},
    {"role": "assistant", "content": chat_response},
    {"role": "user", "content": "What's my name?"}
]

print(messages)

response = llama.chat.completions.create(model="llama3.2", messages=messages)
chat_response = response.choices[0].message.content

print(chat_response)
```
