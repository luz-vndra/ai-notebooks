# Using OpenAI with API Key 

```python
# imports

import os
from dotenv import load_dotenv

from openai import OpenAI
```

```python
# load API key from .env file

load_dotenv(override=True)

api_key = os.getenv('OPENAI_API_KEY')
```

```python
# Check the key

if not api_key:
    print("No API key was found - please head over to the troubleshooting notebook in this folder to identify & fix!")
elif not api_key.startswith("sk-proj-"):
    print("An API key was found, but it doesn't start sk-proj-; please check you're using the right key")
elif api_key.strip() != api_key:
    print("An API key was found, but it looks like it might have space or tab characters at the start or end - please remove them")
else:
    print("API key found and looks good so far!")
```

```python
# build the message to send to the model

message = "Hello, GPT! This is my first ever message to you! Hi!"

messages = [{"role": "user", "content": message}]

messages
```

```python
# generate response from the model

openai = OpenAI() # uses the loaded API key from the environment variable OPENAI_API_KEY

response = openai.chat.completions.create(model="gpt-5-nano", messages=messages) # generate a response from the model using the message built

response.choices[0].message.content 
```
