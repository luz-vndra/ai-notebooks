# 3-way conversation between agents


```python
# imports

import json

from openai import OpenAI
MODEL_LLAMA = "llama3.2"

from IPython.display import display, Markdown
```

```python
# initialize the OpenAI client with Ollama API
agent = OpenAI(base_url="http://localhost:11434/v1", api_key="ollama")
```

```python
# setup system prompt for each persona

alex_system_prompt = """
You are Alex, an eternal optimist. You always see the bright side of things and believe even \
simple actions have deep purpose. Keep replies under 3 sentences.
"""   

blake_system_prompt = """
You are Blake, a witty skeptic who questions everything. You tend to doubt grand explanations \
and prefer clever, sarcastic, or literal answers. Keep replies under 3 sentences.
"""

cameron_system_prompt = """
You are Cameron, a thoughtful philosopher. You consider all perspectives and enjoy finding \
symbolic or existential meaning in simple actions. Keep replies under 3 sentences.
"""
```

```python
# Conversation history
chat_history = [
    {
        "persona": "Alex",
        "text": "Hi, I'm Alex, the eternal optimist."
    }, 
    {
        "persona": "Blake",
        "text": "Hello, I'm Blake, the witty skeptic."
    }, 
    {
        "persona": "Cameron",
        "text": "Greetings, I'm Cameron, the thoughtful philosopher."
    }
]
```

```python
# alex speaks

def alex_speaks():
    prompt = f"The conversation so far: "
    prompt += json.dumps(chat_history)

    response = agent.chat.completions.create(
        model=MODEL_LLAMA,
        messages=[
            {"role":"system", "content": alex_system_prompt},
            {"role": "user", "content": prompt}
        ]
    )


    chat_history.append({ "persona":"Alex", "text": response.choices[0].message.content })
```

```python
alex_speaks()

chat_history
```

```python
# blake speaks

def blake_speaks():
    prompt = f"The conversation so far: "
    prompt += json.dumps(chat_history)

    response = agent.chat.completions.create(
        model=MODEL_LLAMA,
        messages=[
            {"role":"system", "content": blake_system_prompt},
            {"role": "user", "content": prompt}
        ]
    )

    chat_history.append({ "persona":"Blake", "text": response.choices[0].message.content })
```

```python
blake_speaks()
chat_history
```

```python
# cameron speaks

def cameron_speaks():
    prompt = f"The conversation so far: "
    prompt += json.dumps(chat_history)

    response = agent.chat.completions.create(
        model=MODEL_LLAMA,
        messages=[
            {"role":"system", "content": cameron_system_prompt},
            {"role": "user", "content": prompt}
        ]
    )

    chat_history.append({ "persona":"Cameron", "text": response.choices[0].message.content })
```

```python
cameron_speaks()

chat_history
```

```python
import random

for i in range(5):
    random_persona = random.randint(0,2)

    if random_persona == 0:
        alex_speaks()
    elif random_persona == 1:
        blake_speaks()
    elif random_persona == 2:
        cameron_speaks()
```
